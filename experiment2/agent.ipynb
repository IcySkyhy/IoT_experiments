```python
import zmq
import sys
import time
from random import randint
```

```python
numofagents = 2
id = 2
brokerip = '10.0.0.116'
```

```python
def main(id):
    context = zmq.Context()
    publisher = context.socket(zmq.PUB)
    publisher.connect("tcp://%s:5556" % brokerip)

    subscriber = context.socket(zmq.SUB)
    subscriber.connect("tcp://%s:5555" % brokerip)
    subscriber.setsockopt_string(zmq.SUBSCRIBE, str(id))
    while True:
        message = subscriber.recv()
        print('%d: %s' % (id, message))
        # add code for the sensors and LED HAT




        time.sleep(.5)
        publisher.send_string('%d: %s' % (randint(1, numofagents), 'activate'))
```

```python
main(id)
```
